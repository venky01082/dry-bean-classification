This notebook performs exploratory data analysis, data quality checks, feature distribution analysis, and correlation analysis on the UCI Dry Bean Dataset (13,611 samples, 16 features, 7 varieties).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/dry_beans.csv')
print('Shape:', df.shape)
df.head()
```

```python
print('Data Types and Missing Values:')
df.info()
print('\nMissing values count:', df.isnull().sum().sum())
print('Duplicate rows count:', df.duplicated().sum())
```

```python
# Statistical description
df.describe().T
```

```python
# Class distribution
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Class', order=df['Class'].value_counts().index, palette='viridis')
plt.title('Dry Bean Class Distribution (N=13,611)', fontsize=14, fontweight='bold')
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(14, 10))
feats = [c for c in df.columns if c != 'Class']
sns.heatmap(df[feats].corr(), annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Correlation Matrix of 16 Morphological Features', fontsize=14, fontweight='bold')
plt.show()
```
